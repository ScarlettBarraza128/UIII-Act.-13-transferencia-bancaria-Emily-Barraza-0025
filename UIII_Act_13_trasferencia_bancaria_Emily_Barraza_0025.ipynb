### Problema 7
## transferencia bancaria
# atributos
* numeroTransferencia.
* nombreTitular.
* cuentaDestino.
* montoTransferencia.
* saldoDisponible: dinero disponible en la cuenta.

# funciones
* agregarSaldo().
* realizarTransferencia().
* consultarSaldo().
* cancelarTransferencia().
* mostrarDatosTransferencia().

```python
# =========================================================
# IA DE TRANSFERENCIAS BANCARIAS
# Ejemplo de Programación Orientada a Objetos
# Clase: TransferenciaBancaria
# =========================================================

class TransferenciaBancaria:

    # Lista general de transferencias
    lista_transferencias = []

    # =====================================================
    # CONSTRUCTOR
    # =====================================================

    def __init__(self, numeroTransferencia, nombreTitular,
                 cuentaDestino, montoTransferencia,
                 saldoDisponible):

        self.numeroTransferencia = numeroTransferencia
        self.nombreTitular = nombreTitular
        self.cuentaDestino = cuentaDestino
        self.montoTransferencia = montoTransferencia
        self.saldoDisponible = saldoDisponible

    # =====================================================
    # AGREGAR SALDO
    # =====================================================

    def agregarSaldo(self):

        print("\n====== AGREGAR SALDO ======")

        cantidad = float(input("¿Cuánto dinero deseas agregar?: $"))

        self.saldoDisponible = self.saldoDisponible + cantidad

        print("Saldo agregado correctamente.")
        print("Nuevo saldo disponible: $", self.saldoDisponible)

    # =====================================================
    # REALIZAR TRANSFERENCIA
    # =====================================================

    def realizarTransferencia(self):

        print("\n====== REALIZAR TRANSFERENCIA ======")

        if self.montoTransferencia <= self.saldoDisponible:

            self.saldoDisponible = self.saldoDisponible - self.montoTransferencia

            print("Transferencia realizada con éxito.")
            print("Se enviaron: $", self.montoTransferencia)
            print("Cuenta destino:", self.cuentaDestino)
            print("Saldo restante: $", self.saldoDisponible)

        else:
            print("Saldo insuficiente para realizar la transferencia.")

    # =====================================================
    # CONSULTAR SALDO
    # =====================================================

    def consultarSaldo(self):

        print("\n====== CONSULTAR SALDO ======")
        print("Titular:", self.nombreTitular)
        print("Saldo disponible: $", self.saldoDisponible)

    # =====================================================
    # CANCELAR TRANSFERENCIA
    # =====================================================

    def cancelarTransferencia(self):

        print("\n====== CANCELAR TRANSFERENCIA ======")

        print("La transferencia ha sido cancelada.")
        self.montoTransferencia = 0

    # =====================================================
    # MOSTRAR DATOS
    # =====================================================

    def mostrarDatosTransferencia(self):

        print("\n================ DATOS DE LA TRANSFERENCIA ================")

        print("Número de transferencia :", self.numeroTransferencia)
        print("Nombre del titular      :", self.nombreTitular)
        print("Cuenta destino          :", self.cuentaDestino)
        print("Monto transferido       : $", self.montoTransferencia)
        print("Saldo disponible        : $", self.saldoDisponible)

    # =====================================================
    # AGREGAR TRANSFERENCIA
    # =====================================================

    @classmethod
    def agregar_transferencia(cls):

        print("\n========= NUEVA TRANSFERENCIA =========")

        numero = input("Número de transferencia: ")
        titular = input("Nombre del titular: ")
        destino = input("Cuenta destino: ")
        monto = float(input("Monto a transferir: $"))
        saldo = float(input("Saldo disponible: $"))

        nueva = TransferenciaBancaria(
            numero,
            titular,
            destino,
            monto,
            saldo
        )

        cls.lista_transferencias.append(nueva)

        print("\nTransferencia registrada correctamente.")

    # =====================================================
    # LISTAR TRANSFERENCIAS
    # =====================================================

    @classmethod
    def listar_transferencias(cls):

        print("\n================ LISTADO DE TRANSFERENCIAS ================")

        if len(cls.lista_transferencias) == 0:
            print("No hay transferencias registradas.")

        else:

            print("-" * 95)

            print("{:<12} {:<20} {:<20} {:<15} {:<15}".format(
                "NÚMERO",
                "TITULAR",
                "CUENTA DESTINO",
                "MONTO",
                "SALDO"
            ))

            print("-" * 95)

            for transferencia in cls.lista_transferencias:

                print("{:<12} {:<20} {:<20} ${:<14} ${:<14}".format(
                    transferencia.numeroTransferencia,
                    transferencia.nombreTitular,
                    transferencia.cuentaDestino,
                    transferencia.montoTransferencia,
                    transferencia.saldoDisponible
                ))

            print("-" * 95)

    # =====================================================
    # BUSCAR TRANSFERENCIA
    # =====================================================

    @classmethod
    def buscar_transferencia(cls, numeroBuscado):

        for transferencia in cls.lista_transferencias:

            if transferencia.numeroTransferencia == numeroBuscado:
                return transferencia

        return None


# =========================================================
# MENÚ PRINCIPAL
# =========================================================

while True:

    print("\n================================================")
    print("        IA DE TRANSFERENCIAS BANCARIAS")
    print("================================================")

    print("1. Registrar transferencia")
    print("2. Listar transferencias")
    print("3. Agregar saldo")
    print("4. Realizar transferencia")
    print("5. Consultar saldo")
    print("6. Cancelar transferencia")
    print("7. Mostrar datos de transferencia")
    print("8. Salir")

    opcion = input("\nSelecciona una opción: ")

    # =====================================================
    # OPCIÓN 1
    # =====================================================

    if opcion == "1":

        TransferenciaBancaria.agregar_transferencia()

    # =====================================================
    # OPCIÓN 2
    # =====================================================

    elif opcion == "2":

        TransferenciaBancaria.listar_transferencias()

    # =====================================================
    # OPCIÓN 3
    # =====================================================

    elif opcion == "3":

        numero = input("Número de transferencia: ")

        transferencia = TransferenciaBancaria.buscar_transferencia(numero)

        if transferencia:
            transferencia.agregarSaldo()

        else:
            print("Transferencia no encontrada.")

    # =====================================================
    # OPCIÓN 4
    # =====================================================

    elif opcion == "4":

        numero = input("Número de transferencia: ")

        transferencia = TransferenciaBancaria.buscar_transferencia(numero)

        if transferencia:
            transferencia.realizarTransferencia()

        else:
            print("Transferencia no encontrada.")

    # =====================================================
    # OPCIÓN 5
    # =====================================================

    elif opcion == "5":

        numero = input("Número de transferencia: ")

        transferencia = TransferenciaBancaria.buscar_transferencia(numero)

        if transferencia:
            transferencia.consultarSaldo()

        else:
            print("Transferencia no encontrada.")

    # =====================================================
    # OPCIÓN 6
    # =====================================================

    elif opcion == "6":

        numero = input("Número de transferencia: ")

        transferencia = TransferenciaBancaria.buscar_transferencia(numero)

        if transferencia:
            transferencia.cancelarTransferencia()

        else:
            print("Transferencia no encontrada.")

    # =====================================================
    # OPCIÓN 7
    # =====================================================

    elif opcion == "7":

        numero = input("Número de transferencia: ")

        transferencia = TransferenciaBancaria.buscar_transferencia(numero)

        if transferencia:
            transferencia.mostrarDatosTransferencia()

        else:
            print("Transferencia no encontrada.")

    # =====================================================
    # OPCIÓN 8
    # =====================================================

    elif opcion == "8":

        print("\nGracias por usar la IA Bancaria.")
        print("Programa finalizado.")
        break

    # =====================================================
    # OPCIÓN INVÁLIDA
    # =====================================================

    else:

        print("\nOpción inválida. Intenta nuevamente.")
```
